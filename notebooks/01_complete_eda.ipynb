# Phase 01 — Complete EDA and Data Audit

This notebook runs the reproducible SCOPE Phase 01 pipeline.

**Rules**
- No ML models are trained here
- Raw CSV is never overwritten
- Date placeholders are not silently parsed
- Inventory fields are treated as unverified augmented data


```
from pathlib import Path
import json
from IPython.display import Markdown, display, Image

from scope.eda import run_phase01_eda
from scope.config.loader import get_project_root

root = get_project_root()
print('Project root:', root)
```

```
# Run full EDA / audit (writes reports, figures, dictionaries)
result = run_phase01_eda(root)
print(json.dumps(result, indent=2))
```

```
# Preview key reports
for key in ['eda_report', 'data_audit', 'data_dictionary', 'feature_dictionary']:
    path = Path(result['reports'][key])
    display(Markdown(f'### {path.name}'))
    text = path.read_text(encoding='utf-8')
    display(Markdown(text[:2500] + ('\n\n...(truncated)...' if len(text) > 2500 else '')))
```

```
# Show generated figures
fig_dir = Path(result['figures_dir'])
for img in sorted(fig_dir.glob('*.png')):
    display(Markdown(f'**{img.name}**'))
    display(Image(filename=str(img)))
```
